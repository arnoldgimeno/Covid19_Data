### BarPlot Representation of Population to Tests Done Ratio

```python
top20_test_ratio = world_data.sort_values(by="TotalTests", ascending=False).head(20)
```

```python
population_test_ratio=top20_test_ratio['Population']/top20_test_ratio['TotalTests']
```

```python
population_test_ratio
```

```python
#Add the ratio into the dataframe (best practice)
top20_test_ratio["population_test_ratio"] = (
    top20_test_ratio["Population"] / top20_test_ratio["TotalTests"]
)

#Reset the index (optional but makes it look neat)
top20_test_ratio = top20_test_ratio.reset_index(drop=True)

#Handle missing or zero tests (to avoid NaN/inf in the future)
top20_test_ratio = top20_test_ratio.dropna(subset=["Population", "TotalTests"])
top20_test_ratio = top20_test_ratio[top20_test_ratio["TotalTests"] > 0]

#Prepare for Presentatin. Ascending by Country/Region (alphabetical)
plot_df = top20_test_ratio.sort_values(by="Country/Region", ascending=True)

#Ascending by population_test_ratio (best for insight)
plot_df = top20_test_ratio.sort_values(by="population_test_ratio", ascending=True)

#Plot (Clean + Professional) Recommended: Horizontal bar chart (easier to read)
import plotly.express as px

fig3_1 = px.bar(
    plot_df,
    y="Country/Region", #this is the Y-axis
    x="population_test_ratio", #this is the X-axis
    orientation="h", #this is for Horizontal Orientation+
    color="Country/Region", #this assigns different colors
    template="plotly_dark",
    title="<b>Population to Tests Done Ratio (Top 20 by Total Tests)</b>"
)

# Let's make the graph a bit taller
fig3_1.update_layout(
    height=600 # 700(nice), 800(taller), 900(very tall)
)
fig3_1.show()
```

```python
fig3_1.write_html(CHARTS_DIR / "fig3_1_bar_top20_total_test.html")
```

```python
#THIS IS ALTERNATIVE AND SIMPLIER CODE
import plotly.express as px

# Create variable top_20 by TotalTests in descending order.
top20 = world_data.sort_values("TotalTests", ascending=False).head(20)

# Remove (delete) any row where Population OR TotalTests is missing.
# So instead of filling the missing value, we simply removed the entire row.
# drop = remove; na = missing value (NaN)
# if either one has missing value, then delete that raw because it'll create error.
top20 = top20.dropna(subset=["Population", "TotalTests"])

# Let's modify the placeholder top20. Create new column "population_test_ratio", 
# then create ratio: Population / TotalTests using the data that already exist
# in the placeholder top20.
# Keep in mind that "top20" modifies only the Top20 Countries while "world_data"
# modifies the entire dataset. Therefore, we always write "top20" in the algorithm.
top20["population_test_ratio"] = top20["Population"] / top20["TotalTests"]

# Sort the new created column "population_test_ratio" (small → big)
top20 = top20.sort_values("population_test_ratio", ascending=True)

# Plot (same look as your chart)
fig3_4 = px.bar(
    top20,
    x="Country/Region",
    y="population_test_ratio",
    color="Country/Region",
    template="plotly_dark",
    title="<b>Population to Tests Done Ratio (Top 20 by Total Tests)</b>"
)
# Let's make the graph a bit taller
fig3_4.update_layout(
    height=600 # 700(nice), 800(taller), 900(very tall)
)
fig3_4.show()
```

# Pie Charts in Donut Shape

```python
import plotly.express as px

'''
# OPTION 1 (better for comparing multiple metrics-- 'TotalDeaths', 'TotalRecovered', 'ActiveCases'
# --for the same group [Top 10 TotalCases] )

# Pick the TOP 10 worst affected countries (highest TotalCases)
top10 = (
    world_data
    .sort_values("TotalCases", ascending=False) # puts biggest cases first
    .head(10) # takes the real top 10
    .fillna(0) # replaces NaN (missing value) with 0 so the chart doesn’t break
)

cases = ["TotalCases", "TotalDeaths", "TotalRecovered", "ActiveCases"]

for metric in cases: # makes 4 donut charts automatically (one for each metric)
    fig = px.pie(
        top10,
        values=metric,
        names="Country/Region",
        template="plotly_dark",
        hole=0.3,
        title=f"<b>Top 10 worst affected countries by {metric} </b>"
    )

# fig.update_traces(textinfo="percent+label") # this is optional if label countries should be part of the percent value in pie
    
    fig.show()
'''

# OPTION 2 (is etter for TRUE Top 110 lists per metric)
metrics = ["TotalCases", "TotalDeaths", "TotalRecovered", "ActiveCases"]

for metric in metrics:
    top10 = (
        world_data
        .dropna(subset=[metric])              # remove missing metric values
        .sort_values(metric, ascending=False) # biggest first
        .head(10)                             # take top 10
    )

    fig9 = px.pie(
        top10,
        values=metric,
        names="Country/Region",
        hole=0.3,
        template="plotly_dark",
        title=f"<b>Top 10 Countries by {metric}</b>"
    )
    fig9.show()
```

## Death to Confirmed ratio

#### This is also known as Case Fatality Rate (CFR). It tells us how many, out of all confirmed COVID cases, resulted in death. The analysis highlighted that a **higher ratio** indicated a more severe outcome per case, whereas a **lower ratio** pointed to a better survival rate.

```python
deaths_to_confirmed=((world_data['TotalDeaths']/world_data['TotalCases']))
fig10_1 = px.bar(world_data,x='Country/Region',y=deaths_to_confirmed)
fig10_1.update_layout(title={'text':"Death to Confirmed ratio of some  worst effected countries",'xanchor':'left'},
    template="plotly_dark")

# Let's make the graph a bit taller
fig10_1.update_layout(
    height=750 # 700(nice), 800(taller), 900(very tall)
)

fig10_1.show()
```

```python
fig10_1.write_html(CHARTS_DIR / "fig10_1_bar_death2confirmed_ratio.html")
```

```python
import pandas as pd
import plotly.express as px

# Load dataset
df = pd.read_csv(DATA_DIR / "worldometer_data.csv")

# Clean country names
df["Country/Region"] = df["Country/Region"].astype(str).str.strip()

# Use existing normalized mortality metric if available
if "Deaths/1M pop" in df.columns:
    df["MortalityRatePer1M"] = pd.to_numeric(df["Deaths/1M pop"], errors="coerce")
else:
    df["Population"] = pd.to_numeric(df["Population"], errors="coerce")
    df["TotalDeaths"] = pd.to_numeric(df["TotalDeaths"], errors="coerce")
    df["MortalityRatePer1M"] = (df["TotalDeaths"] / df["Population"]) * 1_000_000

# Select top 10 countries by mortality rate
top10 = (
    df.dropna(subset=["MortalityRatePer1M", "TotalDeaths"])
      .sort_values("MortalityRatePer1M", ascending=False)
      .head(10)
      .sort_values("MortalityRatePer1M", ascending=True)
)

# Build chart
fig = px.bar(
    top10,
    x="MortalityRatePer1M",
    y="Country/Region",
    orientation="h",
    title="Top 10 Countries by Mortality Rate",
    text="MortalityRatePer1M",
    color="MortalityRatePer1M",
    color_continuous_scale=["#00E5FF", "#7C4DFF", "#FF4081"],
    hover_data={
        "MortalityRatePer1M": ":,.0f",
        "TotalDeaths": ":,.0f",
        "Population": ":,.0f",
        "Country/Region": False,
    },
)

fig.update_traces(
    texttemplate="%{text:,.0f} / 1M",
    textposition="outside",
    hovertemplate=(
        "<b>%{y}</b><br>"
        "Mortality rate: %{x:,.0f} deaths per 1M<br>"
        "Total deaths: %{customdata[1]:,.0f}<br>"
        "Population: %{customdata[2]:,.0f}<extra></extra>"
    ),
    marker_line_width=0
)

fig.update_layout(
    paper_bgcolor="#0b0f19",
    plot_bgcolor="#0b0f19",
    font={"family": "Arial", "color": "white", "size": 13},
    title={"x": 0.05, "xanchor": "left", "font": {"size": 22, "color": "white"}},
    xaxis=dict(
        title="Deaths per 1M population",
        showgrid=True,
        gridcolor="#22304a",
        zeroline=False,
        showline=False,
        color="white",
    ),
    yaxis=dict(
        title="",
        showgrid=False,
        zeroline=False,
        showline=False,
        color="white",
    ),
    coloraxis_showscale=False,
    margin=dict(l=70, r=40, t=70, b=60),
    height=620,
)

fig.write_html(CHARTS_DIR / "top_10_countries_by_mortality_rate_dark.html",
    include_plotlyjs=True,
    full_html=True,
    default_width="100%",
    default_height="620px",
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv(DATA_DIR / "worldometer_data.csv").copy()
df.columns = [c.strip() for c in df.columns]

for col in ["TotalCases", "TotalTests", "Population", "TotalDeaths"]:
    if col in df.columns:
        df[col] = (
            df[col].astype(str)
            .str.replace(",", "", regex=False)
            .str.replace("+", "", regex=False)
            .replace(["nan", "None", ""], pd.NA)
        )
        df[col] = pd.to_numeric(df[col], errors="coerce")

country_col = "Country/Region" if "Country/Region" in df.columns else df.columns[0]
df[country_col] = df[country_col].astype(str).str.strip()

plot_df = df.dropna(subset=["TotalCases", "TotalTests"]).copy()

exclude = {"World", "Total:", "Europe", "North America", "South America", "Asia", "Africa", "Oceania"}
plot_df = plot_df[~plot_df[country_col].isin(exclude)]

if "Population" in plot_df.columns:
    plot_df["CasesPer1M"] = (plot_df["TotalCases"] / plot_df["Population"]) * 1_000_000

fig = px.scatter(
    plot_df,
    x="TotalTests",
    y="TotalCases",
    size="Population" if "Population" in plot_df.columns else None,
    color="CasesPer1M" if "CasesPer1M" in plot_df.columns else "TotalDeaths" if "TotalDeaths" in plot_df.columns else None,
    hover_name=country_col,
    hover_data={
        "TotalTests": ":,.0f",
        "TotalCases": ":,.0f",
        "Population": ":,.0f" if "Population" in plot_df.columns else False,
        "CasesPer1M": ":,.0f" if "CasesPer1M" in plot_df.columns else False,
        "TotalDeaths": ":,.0f" if "TotalDeaths" in plot_df.columns else False,
    },
    color_continuous_scale=["#00E5FF", "#7C4DFF", "#FF4081"],
    title="Tests vs Confirmed Cases by Country",
)

fig.update_traces(
    marker=dict(opacity=0.80, line=dict(width=0.4, color="white"))
)

fig.update_layout(
    paper_bgcolor="#0b0f19",
    plot_bgcolor="#0b0f19",
    font={"family": "Arial", "color": "white", "size": 13},
    title={"x": 0.05, "xanchor": "left", "font": {"size": 22, "color": "white"}},
    xaxis=dict(
        title="Total Tests (log scale)",
        showgrid=True,
        gridcolor="#22304a",
        zeroline=False,
        showline=False,
        color="white",
        type="log",
    ),
    yaxis=dict(
        title="Total Confirmed Cases (log scale)",
        showgrid=True,
        gridcolor="#22304a",
        zeroline=False,
        showline=False,
        color="white",
        type="log",
    ),
    coloraxis_colorbar=dict(
        title=dict(text="Cases / 1M" if "CasesPer1M" in plot_df.columns else "Total Deaths", font=dict(color="white")),
        tickfont=dict(color="white"),
    ),
    margin=dict(l=70, r=40, t=70, b=60),
    height=620,
)

fig.write_html(CHARTS_DIR / "tests_vs_confirmed_cases_scatter_dark.html",
    include_plotlyjs=True,
    full_html=True,
    default_width="100%",
    default_height="620px",
)

fig.show()
```
